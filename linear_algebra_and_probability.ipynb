## linear algebra and probability


```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("winequality-red.csv", sep=";")
```

```python
df.head()
```

1.Load the dataset and handle missing values

```python
df.shape
```

```python
df = df.fillna(df.mean(numeric_only=True))
```

```python
df.isnull().sum()
```

2.Extract alcohol and citric acid as vectors

```python
alcohol = df["alcohol"].values
citric_acid = df["citric acid"].values
```

```python
# Check vector size
print("Alcohol:", alcohol.shape)
print("Citric Acid:", citric_acid.shape)
```

```python
# Show first 10 values
print(alcohol[:10])
print(citric_acid[:10])
```

3.Calculate covariance matrix

```python
alcohol = df["alcohol"].values
density = df["density"].values
```

```python
cov_matrix = np.cov((alcohol, density), rowvar=False)
```

```python
print(cov_matrix)
```

4.Eigenvalues and eigenvectors

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# Sort from largest to smallest
idx = eigenvalues.argsort()[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:,idx]
```

```python
# Get top 2 eigenvalues
print("Top 2 Eigenvalues:")
print(eigenvalues[:2])

# Get their eigenvectors
print("\nCorresponding Eigenvectors:")
print(eigenvectors[:,:2])
```

5.Find the most common wine quality

```python
# Count quality scores
quality_counts = df["quality"].value_counts()

# Find most common quality
most_common = df["quality"].mode()[0]

print("Quality counts:",quality_counts)
print("\nMost common quality:", most_common)
```

```python
# Count wines with most common quality
count = quality_counts[most_common]
total = len(df)

# Calculate probability
probability = count / total

print("Count:", count)
print("Probability:", probability)
print("Percentage:", probability * 100)
```
