```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Titanic dataset
df = sns.load_dataset("titanic")

# Display the first five rows
df.head()
```

```python
# Countplot: Passenger Count by Embarked Port

plt.figure(figsize=(8,6))

sns.countplot(
    data=df,
    x="embarked",
    hue="sex"
)

plt.title("Passenger Count by Embarked Port")
plt.xlabel("Embarked Port")
plt.ylabel("Number of Passengers")

plt.show()
```

#### Observation

- Most passengers boarded from Southampton (S).
- Male passengers were more than female passengers at all embarkation ports.

```python
# Swarmplot: Age Distribution Across Passenger Classes

plt.figure(figsize=(10,6))

sns.swarmplot(
    data=df,
    x="pclass",
    y="age",
    hue="sex"
)

plt.title("Age Distribution Across Passenger Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Age")

plt.show()
```

#### Observation

- Passengers of all ages traveled in each passenger class.
- First-class passengers show a wider range of ages.
- Male passengers are more numerous than female passengers in all classes.

```python
# Pie Chart: Gender Distribution

# Count male and female passengers
gender = df["sex"].value_counts()

plt.figure(figsize=(6,6))

# Create the pie chart
plt.pie(gender, labels=gender.index, autopct="%1.1f%%")

plt.title("Gender Distribution")

plt.show()
```

#### Observation

- Male passengers make up a larger percentage of the Titanic dataset.
- Female passengers account for a smaller percentage of the total passengers.

```python
# Heatmap: Correlation Matrix

# Select numerical columns and calculate their correlation
corr = df[["age", "fare", "pclass", "sibsp", "parch", "survived"]].corr()

plt.figure(figsize=(8,6))

# Create the heatmap
sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Matrix")

plt.show()
```

#### Observation

- Fare has a positive relationship with survival.
- Passenger class has a negative relationship with fare.
- Age has a weak correlation with the other variables.

```python
# Violin Plot: Fare Distribution by Passenger Class

plt.figure(figsize=(8,6))

# Create the violin plot
sns.violinplot(data=df, x="pclass", y="fare", hue="sex")

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")

plt.show()
```

#### Observation

- First-class passengers generally paid higher fares.
- Third-class passengers mostly paid lower fares.
- Fare distribution varies across passenger classes.

```python
# Scatter Plot: Age vs Fare

plt.figure(figsize=(8,6))

# Create the scatter plot
sns.scatterplot(data=df, x="age", y="fare", hue="survived")

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

#### Observation

- There is no strong relationship between age and fare.
- Passengers who paid higher fares had a better chance of survival.
